```python
# ============================================
# SISTEMA INTELIGENTE DE CAJA Y CONTADOR
# ============================================

# ======================================================
# 🏫 CBTis 128
# 📚 EPPL - Espacios de Programación Lógica
# 🤖 Especialidad: Inteligencia Artificial
# 👨‍🏫 Docente: Eliseo Nava
# 👨‍🎓 Alumno: Jonathan Baray NL = 06
# 💻 Proyecto: Caja y Contador IA
# ======================================================

# =========================
# 💰 Lista vacía
# =========================
lista_clientes = []

# =========================
# 📌 Menú principal
# =========================
while True:

    print("\n=========== MENÚ CAJA Y CONTADOR 💵 ===========")
    print("1. Registrar cliente")
    print("2. Analizar compra")
    print("3. Analizar pagos")
    print("4. Mostrar clientes")
    print("5. Salir")

    opcion = int(input("Selecciona una opción: "))

    # ==================================================
    # ➕ REGISTRAR CLIENTE
    # ==================================================

    if opcion == 1:

        nombre = input("Ingrese nombre del cliente: ")
        tipo_cliente = input("Ingrese tipo de cliente: ")
        contador = input("Ingrese nombre del contador: ")
        caja = input("Ingrese tipo de caja: ")
        total = float(input("Ingrese total de compra: "))
        metodo_pago = input("Ingrese método de pago: ")

        lista_clientes.append([
            nombre,
            tipo_cliente,
            contador,
            caja,
            total,
            metodo_pago
        ])

        print("\n✅ Cliente registrado correctamente")

    # ==================================================
    # 📦 ANALIZAR COMPRA
    # ==================================================

    elif opcion == 2:

        total = float(input("Ingrese total de compra: "))

        print("\n=========== ANÁLISIS IA ===========")

        if total >= 5000:
            print("🏆 Compra alta")

        elif total >= 2000:
            print("📈 Compra media")

        else:
            print("📉 Compra baja")

    # ==================================================
    # 💳 ANALIZAR PAGOS
    # ==================================================

    elif opcion == 3:

        metodo_pago = input("Ingrese método de pago: ")

        print("\n=========== ANÁLISIS IA ===========")

        if metodo_pago == "tarjeta":
            print("💳 Pago electrónico detectado")

        elif metodo_pago == "efectivo":
            print("💵 Pago en efectivo")

        else:
            print("📲 Otro método de pago")

    # ==================================================
    # 📋 MOSTRAR CLIENTES
    # ==================================================

    elif opcion == 4:

        print("\n====================================================================================================")
        print(f"{'CLIENTE':<20}{'TIPO':<15}{'CONTADOR':<20}{'CAJA':<15}{'TOTAL':<15}{'PAGO':<15}")
        print("====================================================================================================")

        i = 0

        while i < len(lista_clientes):

            print(f"{lista_clientes[i][0]:<20}"
                  f"{lista_clientes[i][1]:<15}"
                  f"{lista_clientes[i][2]:<20}"
                  f"{lista_clientes[i][3]:<15}"
                  f"{lista_clientes[i][4]:<15}"
                  f"{lista_clientes[i][5]:<15}")

            i += 1

        print("====================================================================================================")

    # ==================================================
    # ❌ SALIR
    # ==================================================

    elif opcion == 5:

        print("\n💻 Programa finalizado")
        break

    # ==================================================
    # ⚠️ OPCIÓN INVÁLIDA
    # ==================================================

    else:

        print("❌ Opción no válida")+}
```
